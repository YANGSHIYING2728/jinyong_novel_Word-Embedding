```python
import os
import jieba
import numpy as np
import pickle

# 设置小说和数据路径
novel_path = "D:/pycharm/文本理解与数据挖掘/data/"
data_path = "D:/pycharm/文本理解与数据挖掘/"

# 加载停用词
def load_stop_words(stop_words_path):
    with open(stop_words_path, 'r', encoding='gbk', errors='ignore') as file:  # 使用 GBK 编码
        stop_words = [line.strip() for line in file.readlines()]
    return stop_words

stop_words = load_stop_words(data_path + "stop_words.txt")
print(f"加载停用词数量：{len(stop_words)}")

# 加载人物名字并添加到jieba词典
def load_people_names(people_names_path):
    with open(people_names_path, 'r', encoding='gbk', errors='ignore') as file:  # 使用 GBK 编码
        people_names = [line.strip() for line in file.readlines()]
    for name in people_names:
        jieba.add_word(name)  # 逐个添加人名
    return people_names

people_names = load_people_names(data_path + "金庸小说全人物.txt")
print(f"加载人物名字数量：{len(people_names)}")

# 读取小说文件并进行分词
def segment_novel(novel_name):
    novel = open(novel_path + novel_name, 'r', encoding='utf-8-sig')
    print(f"处理 {novel_name}...")
    seg_novel = []
    line = novel.readline()
    while line:
        line = line.strip()
        outstr = ''
        words = jieba.cut(line, cut_all=False)
        for word in words:
            if word not in stop_words and word not in ['\t']:
                if word in people_names:
                    outstr += word + " "  # 保留完整的人名
                else:
                    outstr += word + " "
        if outstr.strip():
            seg_novel.append(outstr.strip().split())
        line = novel.readline()
    novel.close()
    print(f"{novel_name} 处理完成，共 {len(seg_novel)} 行")
    return seg_novel

# 获取所有小说文件名
novel_names = [f for f in os.listdir(novel_path) if f.endswith('.txt')]

# 处理所有小说
seg_novels = []
for novel_name in novel_names:
    seg_novel = segment_novel(novel_name)
    seg_novels.extend(seg_novel)

print("-" * 40)
print("所有小说处理完成，共 {} 行".format(len(seg_novels)))

# 保存处理后的文本数据
with open("seg_novels.pkl", "wb") as f:
    pickle.dump(seg_novels, f)

with open("person_names.pkl", "wb") as f:
    pickle.dump(people_names, f)
```
