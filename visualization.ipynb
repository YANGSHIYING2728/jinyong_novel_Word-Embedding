```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import pickle

plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用黑体
plt.rcParams['axes.unicode_minus'] = False  # 正确显示负号

# 加载降维后的人物向量和人物名字
person_vectors_2d = np.load("person_vectors_2d.npy")
person_vectors_3d = np.load("person_vectors_3d.npy")
with open("person_names.pkl", "rb") as f:
    person_names = pickle.load(f)

# 二维可视化
plt.figure(figsize=(10, 8))
for i, name in enumerate(person_names):
    if i < person_vectors_2d.shape[0]:  # 确保索引不超出范围
        plt.scatter(person_vectors_2d[i, 0], person_vectors_2d[i, 1], label=name)
        plt.text(person_vectors_2d[i, 0], person_vectors_2d[i, 1], name, fontsize=8)
plt.title("2D PCA of Jin Yong Characters")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend()
plt.show()

# 三维可视化
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
for i, name in enumerate(person_names):
    if i < person_vectors_3d.shape[0]:  # 确保索引不超出范围
        ax.scatter(person_vectors_3d[i, 0], person_vectors_3d[i, 1], person_vectors_3d[i, 2], label=name)
        ax.text(person_vectors_3d[i, 0], person_vectors_3d[i, 1], person_vectors_3d[i, 2], name, fontsize=8)
ax.set_title("3D PCA of Jin Yong Characters")
ax.set_xlabel("PCA Component 1")
ax.set_ylabel("PCA Component 2")
ax.set_zlabel("PCA Component 3")
plt.legend()
plt.show()
```
